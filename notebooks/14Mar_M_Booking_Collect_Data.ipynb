```python
from threading import Thread
from time import sleep
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from deep_translator import GoogleTranslator
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
driver = webdriver.Firefox()

# driver.get("https://www.booking.com/hotel/pt/the-yeatman.en-gb.html#tab-reviews")
driver.get("https://www.booking.com/hotel/pt/domhenrique.en-gb.html#tab-reviews")
title = driver.title
sleep(10)
#driver.implicitly_wait(10)
driver.find_element(by=By.ID,value='onetrust-accept-btn-handler').click()


# Step 1: Create an empty DataFrame with five columns
df = pd.DataFrame(columns=['name', 'country', 'room_type', 'nr_nights', 'date', 'traveler_type', 'title_review', 'pos_review', 'neg_review', 'score'])

def call_func(df):
    try:
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.CLASS_NAME, "d799cd346c")))
        mylst = driver.find_elements(By.CLASS_NAME, value="d799cd346c")
        for elem in mylst:
            def get_text(xpath):
                try:
                    return elem.find_element(by=By.XPATH, value=xpath).text
                except Exception:
                    return ''
            x1 = get_text(".//div[@class='a3332d346a e6208ee469']")
            x2 = get_text(".//span[@class='afac1f68d9 a1ad95c055']")
            x3 = get_text(".//span[@data-testid='review-room-name']")
            x4 = get_text(".//span[@data-testid='review-num-nights']")
            x5 = get_text(".//span[@data-testid='review-stay-date']")
            x6 = get_text(".//span[@data-testid='review-traveler-type']")
            x7 = get_text(".//h3[@data-testid='review-title']")
            x8 = get_text(".//div[@data-testid='review-positive-text']")
            x9 = get_text(".//div[@data-testid='review-negative-text']")
            x10 = get_text(".//div[@class='a3b8729ab1 d86cee9b25']").split('\n')[0].replace(',', '.')
            new_row = {'name': x1, 'country': x2, 'room_type': x3, 'nr_nights': x4, 'date': x5, 'traveler_type': x6, 'title_review': x7, 'pos_review': x8, 'neg_review': x9, 'score': x10}
            new_row_df = pd.DataFrame([new_row])
            df = pd.concat([df, new_row_df], ignore_index=True)
        return df
    except Exception as e:
        print(f"Error scraping reviews: {e}")
        return df

number = 0
max = 505

while(number<max):
    df=call_func(df)
    try:
        next_button = WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//button[@aria-label='Next page']")))
        next_button.click()
        number += 1
    except Exception as e:
        print(f"Stopped at page {number}, possibly no more reviews. Error: {e}")
        break

driver.quit()
```
