```python
df.info()
```

```python
df['title_review_en'] = df['title_review_en'].astype(str)
```

```python
df['Token'] = df['title_review_en'].apply(lambda x: word_tokenize(x.lower()))
df
```

```python
stopwords = stopwords.words('english')
punctuation = list(punctuation)
words = set(nltk.corpus.words.words())   
df['Token'] = df['Token'].apply(lambda x: [t for t in x if t not in stopwords and t not in punctuation and t in words])
df
```

```python
df['freqdist'] = df['Token'].apply(lambda x: FreqDist(x))
df
```

```python
#create a function that applies the stemming task
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
def stem_words(text):
    return " ".join([stemmer.stem(word) for word in text])
df["Token"] = df["Token"].apply(lambda x: stem_words(x))
```

```python
#create a function that applies the lemmatizer task
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text])
df["Token"] = df["Token"].apply(lambda text: lemmatize_words(text))
```

```python
df['Token']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df.Token)
df_bow_sklearn = pd.DataFrame(X.toarray(),columns=vectorizer.get_feature_names_out())
df_bow_sklearn
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
X = tfidf_vectorizer.fit_transform(df.Token)
df_bow_sklearn = pd.DataFrame(X.toarray(),columns=tfidf_vectorizer.get_feature_names_out())
df_bow_sklearn
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))
X = bigram_vectorizer.fit_transform(df.Token)
df_bow_sklearn = pd.DataFrame(X.toarray(),columns=bigram_vectorizer.get_feature_names_out())
df_bow_sklearn
```

```python
#df_bow_sklearn['Sentiment']=df['Sentiment']
```

```python
X = df_bow_sklearn.loc[:, df_bow_sklearn.columns != 'Sentiment']
y = df_bow_sklearn.loc[:, df_bow_sklearn.columns == 'Sentiment']
```

```python
clf = DecisionTreeClassifier(max_depth=None, min_samples_split=2, random_state=0)
scores = cross_val_score(clf, X, y, cv=5)
print(scores.mean())

clf = RandomForestClassifier(n_estimators=10, random_state=0)
scores = cross_val_score(clf, X, y, cv=5)
print(scores.mean())
```
