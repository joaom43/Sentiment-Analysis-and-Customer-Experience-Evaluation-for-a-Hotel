```python
import pandas as pd
from unidecode import unidecode
```

```python
# Reads the dataset file: csv.
# Read translated xlsx
df = pd.read_csv("reviews_translated_csv.csv")
```

```python
df.head()
```

```python
# 1) Extrair apenas o número de noites
df['nr_nights'] = df['nr_nights'].str.extract(r'(\d+)')
df['nr_nights'] = pd.to_numeric(df['nr_nights'])

# 2) Separar mês e ano na coluna date
df[['month', 'year']] = df['date'].str.extract(r'(\w+) (\d{4})')
df['year'] = pd.to_numeric(df['year'])

# 3) Extrair apenas o número da coluna score (pode ser inteiro ou decimal)
df['score'] = df['score'].str.extract(r'(\d+\.?\d*)').astype(float)

df['country'].fillna('Unknown', inplace=True)
```

```python
df[['pos_review', 'neg_review', 'pos_review_en', 'neg_review_en']] = df[[
    'pos_review', 'neg_review', 'pos_review_en', 'neg_review_en']].fillna('No review')

# Ter "Nothing" na review quer dizer que a pessoa escreveu "Nothing", quando temos "No review", foi o código a corrigir os NaN.
```

```python
col_order = [
    'name', 'country', 'room_type', 'traveler_type', 'nr_nights', 'date', 'year', 'month', 'score',
    'title_review', 'title_review_en', 'pos_review', 'pos_review_en', 'neg_review', 'neg_review_en']
df = df[col_order]
df.head(2)
```

```python
def clean_text(text):
    if isinstance(text, str):
        text = unidecode(text)  # Remove acentos
        text = text.replace('\n', ' ')  # Substituir quebras de linha
        text = text.strip()  # Remover espaços extra
    return text

# Aplicar a limpeza nas colunas de texto
cols_to_clean = ['title_review', 'pos_review', 'neg_review', 'title_review_en', 'pos_review_en', 'neg_review_en']
df[cols_to_clean] = df[cols_to_clean].applymap(clean_text)
```

```python
for col in df.columns:
    print(f"Valores únicos na coluna '{col}':\n", df[col].unique(), "\n")
```

```python
col_keep = [
    'country', 'room_type', 'traveler_type', 'nr_nights', 'date', 'year', 'month', 'score',
    'title_review_en', 'pos_review_en', 'neg_review_en']
df = df[col_keep]
df.head(2)
```

```python
# Saves the dataframes to .csv and .xlsx
df.to_excel("reviews_clean.xlsx", index=False, encoding="utf-8")
df.to_csv("reviews_clean_csv.csv", index=False, encoding="utf-8")
```
